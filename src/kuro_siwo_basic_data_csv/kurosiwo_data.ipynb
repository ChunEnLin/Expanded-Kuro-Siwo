```python
# 這是從 aoi 中提取 bounding box
def extract_bbox(input_DIR, target_name, output_path):
    DIR = Path(input_DIR)
    # rglob = recursive glob（會往所有子資料夾找）
    # catalogue_files = sorted(DIR.rglob(target_name))
    # catalogue_files = sorted(DIR.glob(f"**/{target_name}"))
    raw_files = DIR.glob(f"*/*/*/{target_name}")
    target_files = sorted([
        p for p in raw_files 
        if not any(part.startswith("EMSR") for part in p.parts)
    ])

    records = []

    for file in target_files:
        gdf = gpd.read_file(file)
        minx, miny, maxx, maxy = gdf.total_bounds

        records.append({
            "filename": str(file),
            "minx": minx,
            "miny": miny,
            "maxx": maxx,
            "maxy": maxy
        })

    df = pd.DataFrame(records)
    df.to_csv(output_path, index=False)


input_DIR = "/mnt/hdd/KuroSiwo_data/slc"
target_name = "aoi.shp"
output_path = "/home/chunen/nas/bigdata/kurosiwo/kurosiwo_bboxes_aoi.csv"

extract_bbox(input_DIR, target_name, output_path)
```

```python
import pandas as pd
from pyproj import Transformer

# 預先定義轉換器 (3857 -> 4326)
transformer = Transformer.from_crs("EPSG:3857", "EPSG:4326", always_xy=True)

def process_coordinates(row):
    """
    這是一個針對每一列 (row) 運行的函數
    """
    mx, my = row['minx'], row['miny']
    MAx, MAy = row['maxx'], row['maxy']
    
    # 判斷這一行的座標是否看起來像 3857 (數值大於 180)
    # 只要 minx 或 maxx 其中一個超出經度範圍，就執行轉換
    if abs(mx) > 180 or abs(MAx) > 180:
        # 轉換左下角
        west, south = transformer.transform(mx, my)
        # 轉換右上角
        east, north = transformer.transform(MAx, MAy)
    else:
        # 如果本來就是 4326，則直接取值
        west, south, east, north = mx, my, MAx, MAy
        
    # 回傳四個新欄位的數值
    return pd.Series([west, south, east, north])

def convert_csv_row_by_row(input_file, output_file):
    # 讀取資料
    df = pd.read_csv(input_file)

    print("正在逐行分析並轉換座標...")

    # 使用 .apply(..., axis=1) 來對「每一列」執行函數
    # 這會產生四個新欄位：west, south, east, north
    df[['west', 'south', 'east', 'north']] = df.apply(process_coordinates, axis=1)

    # 儲存結果
    df.to_csv(output_file, index=False)
    print(f"處理完成！共處理 {len(df)} 筆資料。")
    print(f"結果已儲存至: {output_file}")

# 使用
file_path = "/home/chunen/nas/bigdata/kurosiwo/kurosiwo_bboxes_aoi_revised.csv"
output_path = "/home/chunen/nas/bigdata/kurosiwo/kurosiwo_bboxes_aoi_4326.csv"
convert_csv_row_by_row(file_path, output_path)
```

#### extract_time

```python
def parse_datetime(path: Path) -> tuple[datetime, datetime]:
    parts      = path.name.split("_")
    date1_field = parts[4][0:8]
    date1_timestep = parts[4][9:15]
    date2_field = parts[5][0:8]
    date2_timestep = parts[5][9:15]
    time1 = datetime.strptime(f"{date1_field}{date1_timestep}", "%Y%m%d%H%M%S").replace(tzinfo=timezone.utc)
    time2 = datetime.strptime(f"{date2_field}{date2_timestep}", "%Y%m%d%H%M%S").replace(tzinfo=timezone.utc)
    return time1, time2

def extract_time(input_DIR, target_name, output_path):
    DIR = Path(input_DIR)
    # rglob = recursive glob（會往所有子資料夾找）
    # catalogue_files = sorted(DIR.rglob(target_name))
    raw_folders = DIR.glob(f"*/{target_name}")
    target_folders = sorted([p for p in raw_folders if not p.parent.name.startswith("EMSR")])

    records = []

    for folder in target_folders:
        try:
            time1, time2 = parse_datetime(folder)

            records.append({
                "filename": str(folder),
                "time1": time1,
                "time2": time2
            })
        except Exception as e:
            # 加入例外處理，如果遇到命名不合規的檔案才不會整個程式終止
            print(f"解析跳過 {folder.name}，原因: {e}")

    df = pd.DataFrame(records)
    df.to_csv(output_path, index=False)


input_DIR = "/mnt/hdd/KuroSiwo_data/slc"
target_name = "S1*"
output_path = "/home/chunen/nas/bigdata/kurosiwo/kurosiwo_times.csv"

extract_time(input_DIR, target_name, output_path)
```
